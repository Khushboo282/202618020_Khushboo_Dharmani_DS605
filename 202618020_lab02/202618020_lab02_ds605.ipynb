### Task 2 - Vectorized Arithmetic and Linear Algebra

```python
# Two matrices
M1 = np.array([[1, 2], [3, 4]])
M2 = np.array([[5, 6], [7, 8]])

print("M1:\n", M1)
print("M2:\n", M2)

# Addition
print("\nAddition (M1 + M2):\n", M1 + M2)

# Element-wise multiplication
print("\nElement-wise multiplication (M1 * M2):\n", M1 * M2)

# Matrix multiplication
print("\nMatrix multiplication (M1 @ M2):\n", M1 @ M2)
print("Same via np.matmul:\n", np.matmul(M1, M2))
```

```python
# Transpose and determinant of a square matrix
square = np.array([[4, 7], [2, 6]])
print("Square matrix:\n", square)

print("\nTranspose:\n", square.T)

det = np.linalg.det(square)
print("\nDeterminant:", det)
```

```python
# Inverse (only possible because det != 0) + verification
if not np.isclose(det, 0):
    inv = np.linalg.inv(square)
    print("Inverse:\n", inv)

    identity_check = square @ inv
    print("\nsquare @ inverse:\n", identity_check)
    print("Is it (numerically) the identity matrix?", np.allclose(identity_check, np.eye(2)))
else:
    print("Matrix is singular (det = 0); no inverse exists.")
```

## Task 3 — Normal Distribution and Histogram

```python
# Choose population parameters
pop_mean, pop_std = 50, 10
n_samples = 2000

np.random.seed(1)
normal_data = np.random.normal(loc=pop_mean, scale=pop_std, size=n_samples)

sample_mean = np.mean(normal_data)
sample_std = np.std(normal_data)

print(f"Chosen population mean: {pop_mean}, population std: {pop_std}")
print(f"Sample mean:            {sample_mean:.3f}")
print(f"Sample std:             {sample_std:.3f}")
print(f"Difference in mean:     {abs(sample_mean - pop_mean):.3f}")
print(f"Difference in std:      {abs(sample_std - pop_std):.3f}")
print("\n-> With a large enough sample, the sample statistics converge closely to the",
      "population parameters we chose.")
```

```python
plt.figure(figsize=(8, 5))
plt.hist(normal_data, bins=40, color="steelblue", edgecolor="white", alpha=0.85)
plt.axvline(sample_mean, color="red", linestyle="--", label=f"Sample mean = {sample_mean:.2f}")
plt.title(f"Histogram of {n_samples} samples ~ N({pop_mean}, {pop_std}²)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.savefig("fig_normal_hist.png", dpi=120)
plt.show()
```

### Task 5 — Filtering and Querying

```python
# Q1: Male passengers older than 50
q1 = df[(df["Sex"] == "male") & (df["Age"] > 50)]
print("Male passengers older than 50:", len(q1))
```

```python
# Q2: Female first-class passengers -> count and survival percentage
q2 = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]
q2_survival_pct = q2["Survived"].mean() * 100
print(f"Female first-class passengers: {len(q2)}")
print(f"Survival percentage: {q2_survival_pct:.2f}%")
```

```python
# Q3: Age 20-40, Fare above overall median, and survived
fare_median = df["Fare"].median()
q3 = df.query("20 <= Age <= 40 and Fare > @fare_median and Survived == 1")
print(f"Overall median fare: {fare_median}")
print("Age 20-40, Fare > median, survived:", len(q3))
```

```python
# Q4: Travelling alone (SibSp=0, Parch=0), age < 30, and did NOT survive
q4 = df[(df["SibSp"] == 0) & (df["Parch"] == 0) & (df["Age"] < 30) & (df["Survived"] == 0)]
print("Alone, age < 30, did not survive:", len(q4))
```

```python
# Q5: Embarked='S', Pclass 2 or 3, and Fare above Southampton (S) median
southampton_median_fare = df.loc[df["Embarked"] == "S", "Fare"].median()
q5 = df.query("Embarked == 'S' and Pclass in [2, 3] and Fare > @southampton_median_fare")
print(f"Southampton median fare: {southampton_median_fare}")
print("Embarked=S, Pclass 2/3, Fare > Southampton median:", len(q5))
```

### Task 6 — groupby() and Aggregation

```python
# Survival rate by Sex
survival_by_sex = df.groupby("Sex")["Survived"].mean()
print("Survival rate by Sex:\n", survival_by_sex)
```

```python
# Survival rate by Pclass
survival_by_pclass = df.groupby("Pclass")["Survived"].mean()
print("Survival rate by Pclass:\n", survival_by_pclass)
```

```python
# Average Age and Fare by Pclass
age_fare_by_pclass = df.groupby("Pclass")[["Age", "Fare"]].mean()
print("Average Age & Fare by Pclass:\n", age_fare_by_pclass)
```

```python
# Passenger count and survival rate by Sex-Pclass
sex_pclass_group = df.groupby(["Sex", "Pclass"])["Survived"].agg(["count", "mean"])
sex_pclass_group.columns = ["passenger_count", "survival_rate"]
print("Sex x Pclass:\n", sex_pclass_group)
```

```python
# Passenger count, average Fare, and survival rate by Embarked
embarked_group = df.groupby("Embarked").agg(
    passenger_count=("PassengerId", "count"),
    avg_fare=("Fare", "mean"),
    survival_rate=("Survived", "mean"),
)
print("By Embarked:\n", embarked_group)
```

### Task 7 — Missing Values and Fare Outliers

```python
# Missing value count & percentage per column
missing_count = df.isnull().sum()
missing_pct = (df.isnull().mean() * 100).round(2)
missing_summary = pd.DataFrame({"missing_count": missing_count, "missing_pct": missing_pct})
missing_summary = missing_summary[missing_summary["missing_count"] > 0].sort_values(
    "missing_count", ascending=False
)
print(missing_summary)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7, 4))
plt.bar(missing_summary.index, missing_summary["missing_count"], color="indianred")
plt.title("Missing Values per Column")
plt.ylabel("Missing count")
plt.xlabel("Column")
plt.tight_layout()
plt.savefig("fig_missing_values.png", dpi=120)
plt.show()
```

```python
# Fill missing Age with the mean Age (on a copy so we can compare before/after)
df_filled = df.copy()
print("Missing Age BEFORE:", df_filled["Age"].isnull().sum())

age_mean = df_filled["Age"].mean()
df_filled["Age"] = df_filled["Age"].fillna(age_mean)

print("Missing Age AFTER: ", df_filled["Age"].isnull().sum())
```

```python
# Compare different imputation strategies for Age (mean, median, mode, random)
age_mean_val = df["Age"].mean()
age_median_val = df["Age"].median()
age_mode_val = df["Age"].mode()[0]

np.random.seed(0)
non_null_ages = df["Age"].dropna().values
random_fill = np.random.choice(non_null_ages, size=df["Age"].isnull().sum())

age_mean_imputed = df["Age"].fillna(age_mean_val)
age_median_imputed = df["Age"].fillna(age_median_val)
age_mode_imputed = df["Age"].fillna(age_mode_val)

age_random_imputed = df["Age"].copy()
age_random_imputed[age_random_imputed.isnull()] = random_fill

print(f"Mean imputation value:   {age_mean_val:.2f}")
print(f"Median imputation value: {age_median_val:.2f}")
print(f"Mode imputation value:   {age_mode_val:.2f}")
print("Random imputation: fills each missing Age with a randomly drawn *existing* Age value.")

comparison = pd.DataFrame({
    "mean_imputed": age_mean_imputed.describe(),
    "median_imputed": age_median_imputed.describe(),
    "mode_imputed": age_mode_imputed.describe(),
    "random_imputed": age_random_imputed.describe(),
})
comparison
```

```python
# Fare outliers via IQR method
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = df[(df["Fare"] < lower_bound) | (df["Fare"] > upper_bound)]

print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower bound: {lower_bound:.2f}")
print(f"Upper bound: {upper_bound:.2f}")
print(f"Number of Fare outliers: {len(fare_outliers)}")
```

### Task 8 — Features and Pivot Table

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

df[["Name", "SibSp", "Parch", "FamilySize", "IsAlone"]].head()
```

```python
pivot = pd.pivot_table(df, index="Sex", columns="Pclass", values="Survived", aggfunc="mean")
print("Pivot table (mean Survived, Sex x Pclass):\n", pivot.round(3))

# Find the highest and lowest survival groups
stacked = pivot.stack()
highest_group = stacked.idxmax()
lowest_group = stacked.idxmin()

print(f"\nHighest survival group: Sex={highest_group[0]}, Pclass={highest_group[1]} "
      f"-> {stacked.max():.3f}")
print(f"Lowest survival group:  Sex={lowest_group[0]}, Pclass={lowest_group[1]} "
      f"-> {stacked.min():.3f}")
```

### Task 9 — Visualizations and Observations

```python
numeric_cols = ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare", "FamilySize", "IsAlone"]
corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("fig_correlation_heatmap.png", dpi=120)
plt.show()

# Strongest relationships with Survived (excluding Survived itself)
corr_with_survived = corr["Survived"].drop("Survived").sort_values()
print("Strongest NEGATIVE correlation with Survived:",
      corr_with_survived.index[0], f"({corr_with_survived.iloc[0]:.3f})")
print("Strongest POSITIVE correlation with Survived:",
      corr_with_survived.index[-1], f"({corr_with_survived.iloc[-1]:.3f})")
```

```python
survival_rate_sex = df.groupby("Sex")["Survived"].mean()

plt.figure(figsize=(6, 4))
plt.bar(survival_rate_sex.index, survival_rate_sex.values, color=["salmon", "steelblue"])
plt.title("Survival Rate by Sex")
plt.ylabel("Survival rate")
plt.ylim(0, 1)
for i, v in enumerate(survival_rate_sex.values):
    plt.text(i, v + 0.02, f"{v:.2%}", ha="center")
plt.tight_layout()
plt.savefig("fig_survival_by_sex.png", dpi=120)
plt.show()

print(f"Male survival rate:   {survival_rate_sex['male']:.2%}")
print(f"Female survival rate: {survival_rate_sex['female']:.2%}")
```

```python
plt.figure(figsize=(8, 6))
for label, color in [(0, "indianred"), (1, "seagreen")]:
    subset = df[df["Survived"] == label]
    plt.scatter(subset["Age"], subset["Fare"], alpha=0.5, label=f"Survived={label}", color=color)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare, colored by Survival")
plt.legend()
plt.tight_layout()
plt.savefig("fig_age_vs_fare.png", dpi=120)
plt.show()
```

### Observations

1. female passengers survived at a much higher rate than male passengers 
2. 1st class passengers had a noticeably higher survival rate than 3rd class, consistent with cabin location
   and lifeboat access.
3. Fare correlates positively with survival , largely because Fare is a proxy for Pclass —
   higher fares cluster in 1st class.
4. Age has a weak negative correlation with survival — very young passengers (children)
   had somewhat better survival odds, but the overall linear relationship is weak.
5. Most high fares belong to survivors, visible in the Age vs Fare scatter plot, reinforcing
   the class/fare/survival relationship.
6. FamilySize shows a non-linear relationship with survival— passengers traveling completely
   alone (IsAlone=1) and those in very large families both survived at lower rates than
   passengers in small families (2-4 members).
7. Age and Cabin have the most missing data, with Cabin missing for the large majority of
   passengers, which limits how useful that column is without heavy feature engineering.
